# Phase 7 — Create the Cleaned Dataset

This is the final phase of the Data Cleaning & Preparation project.

The previous phases were used to understand, audit, clean, standardize, and validate the dataset.

This phase creates the final cleaned Excel dataset from the validated data and performs a final check on the exported file.

No additional data cleaning or transformation will be performed in this phase.

```python
import pandas as pd
```

## Load the Validated Dataset

The dataset validated in Phase 6 will be loaded before creating the final cleaned dataset.

```python
final_df = pd.read_excel("../data/processed/cleaned_dataset.xlsx")

final_df.head()
```

```python
final_df.shape
```

```python
final_df.dtypes
```

### Starting Point

The validated dataset contains 1,200 records and 14 columns.

The dataset is being used as-is because the required cleaning, duplicate checks, format checks, and final validation were completed in the previous phases.

## Final Pre-Export Validation

Before creating the final output file, the validated dataset will be checked once more to confirm that it is ready for export.

These checks do not modify the dataset. They only confirm that the required validation conditions still hold.

```python
print("Missing values:", final_df.isnull().sum().sum())
print("Duplicate rows:", final_df.duplicated().sum())
print("Duplicate OrderIDs:", final_df["OrderID"].duplicated().sum())
print("Invalid dates:", pd.to_datetime(final_df["Date"], errors="coerce").isnull().sum())
```

```python
pricing_check = (
    final_df["TotalPrice"]
    == (final_df["Quantity"] * final_df["UnitPrice"]).round(2)
).all()

print("Pricing check:", pricing_check)
```

### Pre-Export Status

All required validation checks passed:

- No missing values
- No duplicate rows
- No duplicate OrderIDs
- No invalid dates
- Pricing calculations remain consistent

The dataset is ready to be exported as the final cleaned dataset.

## Create the Final Cleaned Dataset

The validated dataset will now be exported as the final cleaned Excel dataset.

No additional cleaning or transformation is performed during export.

```python
output_path = "../data/processed/cleaned_dataset.xlsx"

final_df.to_excel(output_path, index=False)

print(f"Final cleaned dataset saved to: {output_path}")
```

### Final Output

The cleaned dataset has been exported to:

`data/processed/cleaned_dataset.xlsx`

This file is the final cleaned dataset produced by the project.

## Verify the Exported Dataset

The exported Excel file will be reloaded to confirm that the final output was saved correctly and that the required validation conditions are preserved after export.

```python
exported_df = pd.read_excel(output_path)

exported_df.shape
```

```python
print("Missing values:", exported_df.isnull().sum().sum())
print("Duplicate rows:", exported_df.duplicated().sum())
print("Duplicate OrderIDs:", exported_df["OrderID"].duplicated().sum())
print("Invalid dates:", pd.to_datetime(exported_df["Date"], errors="coerce").isnull().sum())
```

```python
exported_pricing_check = (
    exported_df["TotalPrice"]
    == (exported_df["Quantity"] * exported_df["UnitPrice"]).round(2)
).all()

print("Pricing check:", exported_pricing_check)
```

### Export Verification Result

The exported Excel file was successfully reloaded and verified.

The final dataset contains 1,200 records and 14 columns, with:

- 0 missing values
- 0 duplicate rows
- 0 duplicate OrderIDs
- 0 invalid dates
- A valid pricing calculation check

The final cleaned dataset is ready for delivery.
