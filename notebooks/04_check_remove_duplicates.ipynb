## Phase 4: Check and Remove Duplicates

The purpose of this phase is to check the cleaned dataset for duplicate records and duplicate order IDs.

The duplicate checks will be performed before making any changes. If duplicates are found, they will be reviewed before deciding whether they should be removed.

## Starting Point

The Phase 3 cleaned dataset contains 1,200 records and 14 columns.

Phase 4 will check this dataset for complete duplicate rows and duplicate `OrderID` values. No records will be removed until the duplicate checks have been completed and the affected records, if any, have been reviewed.

## 4.1 Check for Complete Duplicate Rows

The first check looks for records where every column contains the same value as another record.

```python
duplicate_rows_count = clean_df.duplicated().sum()

duplicate_rows_count
```

```python
duplicate_rows = clean_df[
    clean_df.duplicated(keep=False)
]

duplicate_rows
```

### Finding

No complete duplicate rows were found in the Phase 3 cleaned dataset. Therefore, no duplicate rows need to be removed.

## 4.2 Check for Duplicate OrderIDs

`OrderID` is used to identify each order, so each order should have a unique identifier.

```python
duplicate_order_ids_count = clean_df["OrderID"].duplicated().sum()

duplicate_order_ids_count
```

```python
duplicate_order_ids = clean_df[
    clean_df["OrderID"].duplicated(keep=False)
].sort_values("OrderID")

duplicate_order_ids
```

```python
duplicate_order_id_summary = (
    clean_df.groupby("OrderID")
    .size()
    .reset_index(name="count")
    .query("count > 1")
)

duplicate_order_id_summary
```

### OrderID Finding

No duplicate `OrderID` values were found. Each order currently has a unique `OrderID`, so no records need to be removed based on duplicate identifiers.

## 4.3 Final Duplicate Verification

Both duplicate checks returned zero results, so no records were removed during this phase. The dataset will be checked once more to confirm that there are still no duplicate rows or duplicate `OrderID` values.
