## Column Overview

Before starting the cleaning process, I reviewed the columns to understand what each field represents and how it may be used during the later data-quality checks.

| Column | Description |
|---|---|
| `OrderID` | Unique identifier for each order. |
| `Date` | Date on which the order was placed. |
| `CustomerID` | Identifier for the customer associated with the order. |
| `Product` | Product included in the order. |
| `Quantity` | Number of units ordered. |
| `UnitPrice` | Price of one unit of the product. |
| `ShippingAddress` | Address where the order is to be delivered. |
| `PaymentMethod` | Payment method used for the order. |
| `OrderStatus` | Current status of the order. |
| `TrackingNumber` | Shipment tracking identifier. |
| `ItemsInCart` | Number of items recorded in the customer's cart. |
| `CouponCode` | Coupon code associated with the order, if one was used. |
| `ReferralSource` | Source through which the customer was referred. |
| `TotalPrice` | Total value of the order. |
